## 📊 Sobre la Fuente de Datos: ENSAFI 2023

### ¿En qué consiste la encuesta?
La **Encuesta Nacional sobre Salud Financiera (ENSAFI) 2023** es el primer ejercicio estadístico a nivel nacional realizado por el **INEGI** en colaboración con la **CONDUSEF** para medir de manera integral el bienestar financiero de la población mexicana de 18 años y más. 

Más allá de cuantificar únicamente la tenencia de productos bancarios, la encuesta evalúa la capacidad de las personas para gestionar sus finanzas diarias, afrontar imprevistos económicos, cumplir sus metas a futuro y el impacto emocional o físico que el dinero genera en su vida cotidiana.

### Dimensiones principales que evalúa
* **Control y hábitos financieros:** Elaboración de presupuestos, registro de gastos, puntualidad en el pago de deudas y administración de ingresos.
* **Resiliencia económica:** Capacidad para absorber choques financieros (emergencias médicas, pérdida de empleo o gastos inesperados) mediante ahorros propios, redes de apoyo o crédito.
* **Seguridad y libertad financiera:** Percepción de tranquilidad respecto al futuro económico, ahorro para el retiro y margen para tomar decisiones de vida sin restricciones severas de liquidez.
* **Estrés financiero:** Nivel de preocupación por deudas o gastos corrientes, así como manifestaciones físicas y psicológicas derivadas (ansiedad, insomnio, dolores de cabeza o conflictos familiares).

---

### 🗂️ Estructura de las Tablas (Microdatos)
La base de datos sigue un modelo relacional dividido en cuatro tablas principales conectadas mediante llaves únicas de vivienda (`VIV_SEL`), hogar (`HOGAR`) y residente seleccionado (`N_REN`):

| Archivo | Nivel de análisis | Variables principales que contiene |
| :--- | :--- | :--- |
| **`TSDEM.csv`**[cite: 2] | Sociodemográfico | Edad, sexo, nivel educativo aprobado, estado conyugal, condición de actividad económica y factor de expansión (`fac_sel`). |
| **`TMODULO.csv`**[cite: 2] | Individual (18+ años) | Cuestionario central de salud financiera: ahorro formal e informal, deudas, atrasos en pagos, conocimientos financieros, índice de bienestar y nivel de estrés financiero. |
| **`THOGAR.csv`**[cite: 2] | Hogar | Estructura del hogar, ingresos totales, fuentes de aportación económica y recepción de apoyos gubernamentales o remesas. |
| **`TVIVIENDA.csv`**[cite: 2] | Vivienda | Características físicas de la vivienda, servicios básicos y estatus de tenencia (propia, hipotecada, rentada o prestada). |

---

### ⚠️ Consideraciones Técnicas de los Datos
1. **Codificación numérica y valores centinela:** Las respuestas categóricas están almacenadas como enteros; los códigos `9`, `99`, `999` o `9999` representan *"No sabe"* o *"No especificado"* y deben recodificarse como valores nulos (`NaN`) antes del análisis.
2. **Representatividad estadística:** Para inferir resultados a nivel nacional o estatal, las estimaciones descriptivas deben ponderarse utilizando el factor de expansión muestral (`fac_sel`).

## Empezamos a analizar nuestra base de datos

```python
# Paqueterías a usar
from pathlib import Path
import os
import sys
import numpy as np
import pandas as pd
from IPython.display import display

# Qué archivos se necesitan
required_files = (
    "THOGAR.csv",
    "TMODULO.csv",
    "TSDEM.csv",
    "TVIVIENDA.csv",
)


# Leer los datos
def localizar_archivos_ensafi(nombre_carpeta: str = ".") -> dict[str, Path]:
  """Localiza todos los archivos CSV dentro de la carpeta indicada (y sus subcarpetas)

  y devuelve un diccionario con el nombre del archivo (sin extensión) como
  llave
  y su objeto Path como valor.
  """
  ruta_base = Path(nombre_carpeta).resolve()

  if not ruta_base.exists():
    raise FileNotFoundError(f"La carpeta '{ruta_base}' no existe.")

  # Busca todos los archivos .csv de forma recursiva
  archivos_encontrados = {
      archivo.stem.upper(): archivo for archivo in ruta_base.rglob("*.csv")
  }

  if not archivos_encontrados:
    print(f"No se encontraron archivos CSV en: {ruta_base}")
  else:
    print(f"Archivos localizados en '{ruta_base}':")
    for nombre, ruta in archivos_encontrados.items():
      print(f"  - {nombre}: {ruta.name}")

  return archivos_encontrados


rutas = localizar_archivos_ensafi()

# Cargar directamente las dos tablas que necesites usando el diccionario

df_tsdem = pd.read_csv(rutas["TSDEM"])
df_tmodulo = pd.read_csv(rutas["TMODULO"])
df_tvivienda = pd.read_csv(rutas["TVIVIENDA"])
df_thogar = pd.read_csv(rutas["THOGAR"])

print(f"\nTSDEM cargado: {df_tsdem.shape[0]:,} filas")
print(f"\nTMODULO cargado: {df_tmodulo.shape[0]:,} filas")
print(f"\nTVIVIENDA cargado: {df_tvivienda.shape[0]:,} filas")
print(f"\nTHOGAR cargado: {df_thogar.shape[0]:,} filas")
```
